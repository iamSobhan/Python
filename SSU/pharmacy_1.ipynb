**Input and Output Operations in Python**

<p>Python provides simple built-in functions to handle Input (taking data from the user) and Output (displaying data to the user).</p><br>
<p>In Python, the input() function is used to take data from the user, while the print() function is used to display data as output on the screen.</p>

**Output Operation**

print() function sends data to the standard output device (usually our monitor console). It can output strings, numbers, variables, or the results of expressions.

**Basic usage:**

```python
print("Seacom Skills University")
```

**Printing multiple values:**

```python
print("Seacom Skills University", "Birbhum", 731236)
```

```python
print("Hello World!")
print("I am learning Python.")
print("It's a awesome programming language!")
```

**Using separator (sep):**

```python
print("Python", "Java", "C", "C++", sep=" | ")
```

**Using end parameter (default is newline):**

```python
print("Department", end=" ")
print("of", end=" ")
print("Pharmacy")
```

**Formatted output using f-strings**

```python
name = "Iswar Kundu"
roll = 24
print(f"My name is {name} and my roll no is {roll}.")
```

```python
print("This is a Statement.")
print('This is also a Statement.')
```

**Print Numbers**

```python
print(3)
print(75)
print(90000)
```

**Math Operations**

```python
print(5 + 2)
print(2 * 5)
```

**Input Operation**

Used to take input from the user via keyboard. Always returns a string by default.<br>
The input() function pauses program execution and waits for the user to type something using their keyboard. Once the user presses Enter, the function reads the text and passes it into the program.

**Basic usage:**

```python
name = input("Enter your name: ")
print("Hello, I am ", name)
```

**Taking numeric input**

```python
age = int(input("Enter your age: "))
print("My age is ", age)
```

**Combining Input & Output**

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
print(f"Hello I am {name}, I will be {age + 1} next year!")
```

```python
name = input('Enter your Name: ')
age = int(input('Enter your Age: '))
print('Hii I am', name, 'and my age is', age)
```
