**Python Strings**

<p>A string in Python is a sequence of characters enclosed in single (' '), double (" "), or triple (''' ''' / """ """) quotes. Strings are of type str and are immutable, meaning they cannot be changed after creation.<p/>

**Creating Strings**

```python
a = 'Hello'
b = "Python"
c = """This is a
multi-line string"""
print(type(a))
print(a)
print(b)
print(c)
```

**Indexing and Slicing**

<p>Each character has a position (index), starting from 0. Negative indexes count from the end.</p>

```python
s = "Python"
print(s[0])      # P
print(s[-1])     # n
print(s[0:3])    # Pyt  (start included, end excluded)
print(s[2:])     # thon
print(s[::-1])   # nohtyP  (reversed)
print(s[:-1])

print(s[0])
print(s[1])
print(s[2])
print(s[3])
print(s[4])
print(s[5])

print(s[-1])
print(s[-2])
print(s[-3])
print(s[-4])
print(s[-5])
print(s[-6])

print(s[0:1])
print(s[0:2])
print(s[0:3])
print(s[0:4])
print(s[0:5])
print(s[0:6])

```

<p> In Python slicing ([start:end:step]), s[:-1] and s[::-1] do two completely different operations because of how the step parameter works.

1. s[:-1] (Standard Slicing)
Syntax breakdown: [start : end : step] where start defaults to the beginning (0), end is -1 (exclusive), and step defaults to +1.

What it does: It slices the string from the beginning up to, but not including, the last character.

Output: Pytho

Why: It drops just the last character ("n").

2. s[::-1] (Reversing with Step)
Syntax breakdown: [start : end : step] where start defaults to the end, end defaults to the beginning, and step is -1.

What it does: The negative step (-1) tells Python to traverse the string backwards, moving from right to left instead of left to right.

Output: nohtyP

Why: It completely reverses the entire string.</p>

**Case Conversion Methods**

```python
#upper() converts all characters to uppercase.

print("hello".upper())
print("seacom skills university".upper())
print("python programming".upper())
```

```python
#lower() converts all characters to lowercase.

print("HII".lower())
print("DEPARTMENT OF PHARMACY".lower())
print("MACHINE LEARNING".lower())
```

```python
#title() capitalizes the first letter of every word.

print("hello world".title())
print("python programming".title())
print("deep learning".title())
```

```python
#capitalize() capitalizes only the first letter of the string.

print("hello Kolkata".capitalize())
print("python programming".capitalize())
print("deep learning".capitalize())
```

```python
#swapcase() swaps uppercase to lowercase and vice versa.

print("Hello World".swapcase())
print("Python Programming".swapcase())
print("Natural Language Processing".swapcase())
```

**Whitespace Removal Methods**

```python
#strip() removes spaces from both ends.

print("   Hello World   ".strip())
print("   Python Programming   ".strip())
print("   Natural Language Processing   ".strip())
```

```python
#lstrip() removes spaces from the left side only.

print("   Hello World   ".lstrip())
print("   Python Programming   ".lstrip())
print("   Natural Language Processing   ".lstrip())
```

```python
#rstrip() removes spaces from the right side only.

print("   Hello World   ".rstrip())
print("   Python Programming   ".rstrip())
print("   Natural Language Processing   ".rstrip())
```

**Searching Methods**

```python
#find() returns the index of the first occurrence, or -1 if not found.

print("Python".find("th"))
print("Python".find("z"))
```

```python
#index() works like find(), but raises a ValueError if not found.

print("Python".index("th"))
```

```python
# rfind() finds the last occurrence.
print("banana".rfind("a"))
```

```python
# count() counts occurrences of a substring.

print("banana".count("a"))
```

```python
# startswith() checks whether the string begins with a given text.

print("Python".startswith("Py"))
```

```python
#endswith() checks whether the string ends with a given text.

print("Python".endswith("th"))
```

**Modification Methods**

```python
# replace() replaces a substring with another.

print("I like cats".replace("cats", "dogs"))
print("banana".replace("a", "o", 2))
```

```python
#split() splits a string into a list.

print("a,b,c".split(","))
print("I love Python".split())
```

```python
#join() joins items of a list into a string.

print("-".join(["a", "b", "c"]))   # a-b-c
print(" ".join(["Hello", "World"]))
```

```python
#splitlines() splits a string at line breaks.

print("Line1\nLine2\nLine3".splitlines())
```

**Alignment and Padding Methods**

```python
#ljust() left-aligns the string.

print("Hi".ljust(6, "-"))
```

```python
#rjust() right-aligns the string.

print("Hi".rjust(6, "-"))
```

```python
#zfill() pads with zeros on the left.

print("42".zfill(5))
```

**Escape Characters in Python**

<p>An escape character is a backslash (\) followed by a character. It tells Python to treat that character in a special way, such as inserting a new line, a tab, or a quote inside a string.</p>

```python
#\n (New Line)
#Moves the text after it to the next line.

print("Hello\nWorld")
print("Hii\nThis is\na Notebook")

print("This is Line 1\nThis is Line 2\nThis is Line 3")
```

```python
#\t (Horizontal Tab)
#Inserts a tab space, useful for aligned output.

print("Name\tAge\tCity")
print("Chiku\t25\tDelhi")
```

```python
#\\ (Backslash)
#Prints a single backslash.

print("C:\\Users\\Documents")
```

```python
#\' (Single Quote)
#Lets you use a single quote inside a single-quoted string.

print("He said \"Hello\" to me")
```

```python
#\r (Carriage Return)
#Moves the cursor to the start of the line, so later text overwrites earlier text.

print("Hello\rWorld")

#Here, World overwrites Hello because both have 5 characters. (The exact display can vary by terminal.)
```

```python
#\b (Backspace)
#Removes the character immediately before it.

print("Helloo\b World")
```

```python
#\f (Form Feed)
#Inserts a page break. It is rarely used and may appear as a symbol in some terminals.

print("Page 1\fPage 2")
```

```python
#\v (Vertical Tab)
#Inserts a vertical tab. It is also rarely used and may display differently across systems.

print("Hello\vWorld")
```

```python
#\a (Bell / Alert)
#Produces an alert sound on some systems.

print("Alert!\a")
```

```python
#\0 (Null Character)
#Represents the null character (ASCII 0). Mostly used in low-level data handling.

s = "Hello\0World"
print(len(s))
```

```python
#\ooo (Octal Value)
#Inserts a character using its octal ASCII code.

print("\110\145\154\154\157")
```

```python
#\xhh (Hexadecimal Value)
#Inserts a character using its hexadecimal ASCII code.

print("\x48\x65\x6c\x6c\x6f")
```

```python
#\uXXXX (Unicode, 16-bit)
#Inserts a Unicode character using a 4-digit hex code.

print("\u2764")
print("\u03b1")
print("\u00b5g")
```

```python
#\UXXXXXXXX (Unicode, 32-bit)
#Inserts a Unicode character using an 8-digit hex code.

print("\U0001F600")
```

```python
#\N{name} (Unicode by Name)
#Inserts a Unicode character using its official name.

print("\N{DEGREE SIGN}")
print("\N{GREEK SMALL LETTER BETA}")
```

```python
#\ at End of Line (Line Continuation)
#Continues a string across lines in the source code without adding a new line to the output.

text = "This is a long \
sentence in one line"
print(text)
```
