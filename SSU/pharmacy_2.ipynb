**Comments in Python**

A Python comment is a line of text in our code that the Python interpreter ignores during execution. <br>

Comments are used to explain code, leave notes for ourself or other developers, or temporarily disable code. They don't affect how the program runs.

**Single-Line Comments**


```python
# This is a single-line comment
print("Hello, This is the Pharmacy Department!")  # This prints a greeting message.
```

**Multi-Line Comments** <br><br>
**a) Using multiple # symbols**

```python
# This is line one of the comment
# This is line two of the comment
# This is line three of the comment
print("This is a Multi-line comment example.")
```
